```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import torch
```

```python
from skimage.data import camera, coffee, astronaut
```

```python
img = camera()
type(img)
img.shape  # (H, W) grayscale img without channel dimension
```

```python
_ = plt.imshow(img, cmap='gray')
```

```python
img = coffee()
img.shape  # (H, W, C)
```

```python
_ = plt.imshow(img)
```

```python
red = img[:, :, 0]
green = img[:, :, 1]
blue = img[:, :, 2]
```

```python
fig, axes = plt.subplots(1, 3, figsize=(21, 7))
_ = axes[0].imshow(red, cmap='gray')
_ = axes[1].imshow(green, cmap='gray')
_ = axes[2].imshow(blue, cmap='gray')
```

```python
red.min()
red.max()
```

```python
red.mean()
green.mean()
blue.mean()
```

```python
red_normal = red / 255
red_normal.min()
red_normal.max()
_ = plt.imshow(red_normal, cmap='gray')
```

# MNIST dataset

```python
from torchvision.datasets import MNIST
```

```python
train_data = MNIST(root='./data', train=True, download=True)
test_data = MNIST(root='./data', train=False, download=True)
```

```python
len(train_data)
len(test_data)
```

```python
img, label = train_data[0]
img
label
```

```python
_ = plt.imshow(img, cmap='gray')
```

```python
fig, axes = plt.subplots(2, 5, figsize=(10, 5))
axes = axes.flatten()

for i in range(10):
  img, label = train_data[i]
  _ = axes[i].imshow(img, cmap='gray')
  _ = axes[i].set_title(label)
  _ = axes[i].set_axis_off()

plt.tight_layout()
```
